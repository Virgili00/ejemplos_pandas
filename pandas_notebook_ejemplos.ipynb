##Slicing: Selección Avanzada de Datos

Slicing es la técnica para seleccionar subconjuntos de datos (porciones de la tabla). Se usan principalmente `.loc` e `.iloc`.

- **`.loc[filas, columnas]`**: Selección por **ETIQUETA**. El rango final SÍ se incluye.
- **`.iloc[filas, columnas]`**: Selección por **POSICIÓN**. El rango final NO se incluye.

```python
# --- Ejemplos con .iloc (basado en POSICIÓN) ---
print("--- .iloc: Fila en posición 1 ---")
display(df.iloc[1])
```

```python
# Fila en posición 1, columna en posición 2 ('Ventas')
df.iloc[1, 2]
```

```python
# Primeras 3 filas y primeras 2 columnas
df.iloc[0:3, 0:2]
```

```python
df.loc[0:2,'Producto':'Costos']
```

```python
# --- Ejemplos con .loc (basado en ETIQUETA) ---
# Para que sea más claro, asignamos etiquetas de índice personalizadas
df.index = ['Fila_A', 'Fila_B', 'Fila_C', 'Fila_D', 'Fila_E', 'Fila_F', 'Fila_G']

print("--- DataFrame con nuevo índice ---")
display(df.head())

# Fila con la etiqueta 'Fila_B'
print("\n--- .loc: Fila con etiqueta 'Fila_B' ---")
display(df.loc[['Fila_B']])

# Filas desde 'Fila_B' hasta 'Fila_D' y columnas 'Producto' y 'Ventas'
print("\n--- .loc: Slicing de filas y columnas por etiqueta ---")
display(df.loc['Fila_B':'Fila_D', ['Producto', 'Ventas']])
```

##Calculo de columnas

```python
# Ejemplo 1: Calcular la ganancia (Ventas - Costos)
# Se crea una nueva columna llamada 'Ganancia'
df['Ganancia'] = df['Ventas_Totales'] - df['Costos']
df

```

```python
# Ejemplo 2: Calcular el margen de ganancia en porcentaje
# Se crea la columna 'Margen_%'
df['Margen_%'] = (df['Ganancia'] / df['Ventas_Totales']) * 100
print("\n--- DataFrame con columna 'Margen_%' ---")
# Usamos .round(2) para redondear el resultado a 2 decimales
df.round(2)
```

## Uso de `apply()`

El método `apply()` se utiliza para aplicar una función a lo largo de un eje de un DataFrame (filas o columnas) o a los elementos de una Serie.

```python
# Ejemplo 1: Crear una nueva columna 'Clasificacion_Ventas' basada en 'Ventas_Totales' usando apply()
def clasificar_ventas(ventas):
    if ventas > 180:
        return 'Muy Altas'
    elif ventas > 120:
        return 'Altas'
    else:
        return 'Bajas'

df['Clasificacion_Ventas'] = df['Ventas_Totales'].apply(clasificar_ventas)
print("--- DataFrame con 'Clasificacion_Ventas' usando apply() ---")
display(df)
```

```python
# Ejemplo 2: Calcular el 'Beneficio_Unitario' usando apply() con una función lambda
# Asumimos que la ganancia es Ventas_Totales - Costos, para este ejemplo
df['Beneficio_Unitario'] = df.apply(lambda row: (row['Ventas_Totales'] - row['Costos']) / row['Ventas_Totales'] if row['Ventas_Totales'] > 0 else 0, axis=1)
print("\n--- DataFrame con 'Beneficio_Unitario' usando apply() con lambda ---")
display(df[['Ventas_Totales', 'Costos', 'Beneficio_Unitario']].round(2))
```

## Uso de `where()`

El método `where()` se utiliza para reemplazar valores en un DataFrame o Serie donde una condición es `False`. Los valores donde la condición es `True` se mantienen intactos.

```python
# Ejemplo 1: Reemplazar los 'Costos' con 0 donde las 'Ventas_Totales' sean bajas
df_where = df.copy() # Hacemos una copia para no modificar el DataFrame original
df_where['Costos_Ajustados'] = df_where['Costos'].where(df_where['Ventas_Totales'] > 120, 0)
print("--- DataFrame con 'Costos_Ajustados' usando where() ---")
display(df_where[['Ventas_Totales', 'Costos', 'Costos_Ajustados']])

# Ejemplo 2: Reemplazar la 'Region' con 'Desconocida' donde 'Producto' sea 'C'
df_where['Region_Ajustada'] = df_where['Region'].where(df_where['Producto'] != 'C', 'Desconocida')
print("\n--- DataFrame con 'Region_Ajustada' usando where() ---")
display(df_where[['Producto', 'Region', 'Region_Ajustada']])
```

## Uso de `map()`

El método `map()` se utiliza para sustituir cada valor en una Serie con otro valor, lo cual es útil para mapear valores categóricos a números o para aplicar una función de mapeo basada en un diccionario o una Serie.

```python
# Ejemplo 1: Mapear 'Region' a valores numéricos usando un diccionario
regiones_map = {'Norte': 1, 'Sur': 2, 'Este': 3, 'Oeste': 4, 'Desconocida': 0}
df['Region_Mapeada'] = df['Region'].map(regiones_map)
print("--- DataFrame con 'Region_Mapeada' usando map() ---")
display(df[['Region', 'Region_Mapeada']])
```
